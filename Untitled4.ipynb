```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns

# Read the official seasonal agriculture dataset
df = pd.read_csv("seasonal_agriculture_performance_dataset.csv")
display(df.head(10))

# Display complete information
print("\n--- Complete Dataset Info ---")
df.info()

# Summary statistics for numeric columns
print("\n--- Summary Statistics ---")
print(df.describe())

# Check missing values
print("\n--- Missing Values Count ---")
print(df.isnull().sum())

# Visulization and Analyse the data
y = list(df.Profit_INR)
plt.boxplot(y)
plt.show()

# Visualization by Heatmap
sns.heatmap(df.isnull(),cbar=False,cmap="viridis")

# To plot Scatter plot
plt.figure(figsize = (10,10))
plt.scatter (x="Farm_Area_Hectares" , y="Production_Tonnes" , data = df)
plt.xlabel("Farm_Area_Hectares")
plt.ylabel("Production_Tonnes")
```
